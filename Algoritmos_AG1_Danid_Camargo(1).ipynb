##Devolución de cambio por técnica voraz

```python
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION


cambio_monedas(15,[25,10,5,1])
```

## N-Reinas por técnica de vueta atrás


```python
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")


def es_prometedora(SOLUCION,etapa):
  #print(SOLUCION)
  #Si la solución tiene dos valores iguales no es valida => Dos reinas en la misma fila
  for i in range(etapa+1):
    #print("El valor " + str(SOLUCION[i]) + " está " +  str(SOLUCION.count(SOLUCION[i])) + " veces")
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False

    #Verifica las diagonales
    for j in range(i+1, etapa +1 ):
      #print("Comprobando diagonal de " + str(i) + " y " + str(j))
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]) : return False
  return True



def reinas(N, solucion=[], etapa=0):
  if len(solucion) == 0:
      solucion=[0 for i in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        #escribe(solucion)
        print()
      else:
        reinas(N, solucion, etapa+1)
    else:
      None

    solucion[etapa] = 0

reinas(8)
```

## Viaje por el rio. Programación dinámica

```python
TARIFAS = [
[0,5,4,3,999,999,999],
[999,0,999,2,3,999,11],
[999,999, 0,1,999,4,10],
[999,999,999, 0,5,6,9],
[999,999, 999,999,0,999,4],
[999,999, 999,999,999,0,3],
[999,999,999,999,999,999,0]
]



################################################################
def Precios(TARIFAS):
################################################################
  #Total de Nodos
  N = len(TARIFAS[0])

  #Inicialización de la tabla de precios
  PRECIOS = [ [9999]*N for i in [9999]*N]
  RUTA = [ [""]*N for i in [""]*N]

  for i in range(0,N-1):
    RUTA[i][i] = i             #Para ir de i a i se "pasa por i"
    PRECIOS[i][i] = 0          #Para ir de i a i se se paga 0
    for j in range(i+1, N):
      MIN = TARIFAS[i][j]
      RUTA[i][j] = i

      for k in range(i, j):
        if PRECIOS[i][k] + TARIFAS[k][j] < MIN:
            MIN = min(MIN, PRECIOS[i][k] + TARIFAS[k][j] )
            RUTA[i][j] = k          #Anota que para ir de i a j hay que pasar por k
        PRECIOS[i][j] = MIN

  return PRECIOS,RUTA
################################################################


PRECIOS,RUTA = Precios(TARIFAS)
#print(PRECIOS[0][6])

print("PRECIOS")
for i in range(len(TARIFAS)):
  print(PRECIOS[i])

print("\nRUTA")
for i in range(len(TARIFAS)):
  print(RUTA[i])

#Determinar la ruta con Recursividad
def calcular_ruta(RUTA, desde, hasta):
  if desde == hasta:
    #print("Ir a :" + str(desde))
    return ""
  else:
    return str(calcular_ruta( RUTA, desde, RUTA[desde][hasta])) +  \
                ',' + \
                str(RUTA[desde][hasta] \
              )

print("\nLa ruta es:")
calcular_ruta(RUTA, 0,6)
```

6. ejercico encontrar los dos puntos mas cercanos

```python
import math

# ==========================================
# 1. 1D: Primer intento (Fuerza Bruta)
# Complejidad: O(n^2) - Compara todos contra todos.
# ¿Se puede mejorar?: Sí, ordenando la lista previamente.
# ==========================================
def puntos_cercanos_1d_fuerza_bruta(puntos):
    min_dist = float('inf')
    par = None
    n = len(puntos)
    for i in range(n):
        for j in range(i + 1, n):
            dist = abs(puntos[i] - puntos[j])
            if dist < min_dist:
                min_dist, par = dist, (puntos[i], puntos[j])
    return par, min_dist

# ==========================================
# 2. 1D: Segundo intento (Divide y Vencerás)
# Complejidad: O(n log n) debido al ordenamiento inicial.
# ¿Se puede mejorar?: No mediante comparaciones puras; O(n log n) es el límite óptimo.
# ==========================================
def puntos_cercanos_1d_divide_venceras(puntos):
    # El ordenamiento es clave para mejorar la complejidad
    puntos_ordenados = sorted(puntos)
    return _recursividad_1d(puntos_ordenados)

def _recursividad_1d(puntos):
    n = len(puntos)
    if n < 2: return None, float('inf')
    if n == 2: return (puntos[0], puntos[1]), puntos[1] - puntos[0]

    mitad = n // 2
    par_izq, dist_izq = _recursividad_1d(puntos[:mitad])
    par_der, dist_der = _recursividad_1d(puntos[mitad:])

    min_dist = min(dist_izq, dist_der)
    mejor_par = par_izq if dist_izq <= dist_der else par_der

    # Comprobar la frontera entre las dos mitades
    dist_frontera = puntos[mitad] - puntos[mitad - 1]
    if dist_frontera < min_dist:
        min_dist = dist_frontera
        mejor_par = (puntos[mitad - 1], puntos[mitad])

    return mejor_par, min_dist

# ==========================================
# 3. 2D: Extender el algoritmo
# Complejidad: O(n log n) aplicando la partición por franja central.
# ==========================================
def distancia_n_dim(p1, p2):
    return math.dist(p1, p2) # math.dist soporta 2D, 3D y N-dimensiones automáticamente

def fuerza_bruta_general(puntos):
    min_dist, par = float('inf'), None
    for i in range(len(puntos)):
        for j in range(i + 1, len(puntos)):
            d = distancia_n_dim(puntos[i], puntos[j])
            if d < min_dist:
                min_dist, par = d, (puntos[i], puntos[j])
    return par, min_dist

def puntos_cercanos_2d(puntos):
    puntos_x = sorted(puntos, key=lambda p: p[0])
    return _recursividad_2d(puntos_x)

def _recursividad_2d(puntos_x):
    n = len(puntos_x)
    if n <= 3: return fuerza_bruta_general(puntos_x)

    mitad = n // 2
    punto_medio = puntos_x[mitad]

    par_izq, dist_izq = _recursividad_2d(puntos_x[:mitad])
    par_der, dist_der = _recursividad_2d(puntos_x[mitad:])

    d = min(dist_izq, dist_der)
    mejor_par = par_izq if dist_izq <= dist_der else par_der

    # Analizar la franja central (strip)
    franja = [p for p in puntos_x if abs(p[0] - punto_medio[0]) < d]
    franja.sort(key=lambda p: p[1])

    # Comprobación de puntos dentro de la franja (máx 7 comprobaciones matemáticas)
    for i in range(len(franja)):
        for j in range(i + 1, min(i + 7, len(franja))):
            d_franja = distancia_n_dim(franja[i], franja[j])
            if d_franja < d:
                d, mejor_par = d_franja, (franja[i], franja[j])

    return mejor_par, d

# ==========================================
# 4. 3D: Extender el algoritmo
# Aprovechando la generalización de math.dist
# ==========================================
def puntos_cercanos_3d(puntos):
    # Para 3D estricto sin usar librerías avanzadas (como scipy.spatial.KDTree),
    # la distancia euclidiana generalizada resuelve el problema geométrico de la misma forma.
    return fuerza_bruta_general(puntos)

# ==========================================
# Pruebas de Ejecución (Para tu celda)
# ==========================================
if __name__ == "__main__":
    lista_1d = [3403, 4537, 9089, 9746, 7259, 1200, 3400]
    print("--- Resultados 1D ---")
    print(f"Fuerza Bruta: {puntos_cercanos_1d_fuerza_bruta(lista_1d)}")
    print(f"Divide y Vencerás: {puntos_cercanos_1d_divide_venceras(lista_1d)}\n")

    lista_2d = [(1122, 6175), (135, 4076), (7296, 2741), (1120, 6170)]
    print("--- Resultados 2D ---")
    print(f"Divide y Vencerás 2D: {puntos_cercanos_2d(lista_2d)}\n")

    lista_3d = [(10, 20, 30), (12, 22, 31), (100, 200, 300), (11, 21, 30)]
    print("--- Resultados 3D ---")
    print(f"Extensión a 3D: {puntos_cercanos_3d(lista_3d)}")
```

```python
def puntos_cercanos_1d_divide_venceras(puntos):
    # Ordenamiento inicial crucial para optimizar
    puntos_ordenados = sorted(puntos)
    return _recursividad_1d(puntos_ordenados)

def _recursividad_1d(puntos):
    n = len(puntos)
    if n < 2: return None, float('inf')
    if n == 2: return (puntos[0], puntos[1]), puntos[1] - puntos[0]

    mitad = n // 2
    par_izq, dist_izq = _recursividad_1d(puntos[:mitad])
    par_der, dist_der = _recursividad_1d(puntos[mitad:])

    # Encontrar el mínimo entre las dos mitades
    min_dist = min(dist_izq, dist_der)
    mejor_par = par_izq if dist_izq <= dist_der else par_der

    # Verificar la frontera (el último de la izquierda y el primero de la derecha)
    dist_frontera = puntos[mitad] - puntos[mitad - 1]
    if dist_frontera < min_dist:
        min_dist = dist_frontera
        mejor_par = (puntos[mitad - 1], puntos[mitad])

    return mejor_par, min_dist

# Prueba en celda
print("1D Divide y Vencerás:", puntos_cercanos_1d_divide_venceras(puntos_1d))
```

```python
import math

def distancia_euclidiana(p1, p2):
    return math.hypot(p1[0] - p2[0], p1[1] - p2[1])

def fuerza_bruta_2d(puntos):
    min_dist = float('inf')
    par = None
    for i in range(len(puntos)):
        for j in range(i + 1, len(puntos)):
            d = distancia_euclidiana(puntos[i], puntos[j])
            if d < min_dist:
                min_dist, par = d, (puntos[i], puntos[j])
    return par, min_dist

def puntos_cercanos_2d(puntos):
    # Se ordena por la coordenada X
    puntos_x = sorted(puntos, key=lambda p: p[0])
    return _recursividad_2d(puntos_x)

def _recursividad_2d(puntos_x):
    n = len(puntos_x)
    if n <= 3:
        return fuerza_bruta_2d(puntos_x)

    mitad = n // 2
    punto_medio = puntos_x[mitad]

    par_izq, dist_izq = _recursividad_2d(puntos_x[:mitad])
    par_der, dist_der = _recursividad_2d(puntos_x[mitad:])

    d = min(dist_izq, dist_der)
    mejor_par = par_izq if dist_izq <= dist_der else par_der

    # Construir franja central de ancho 2d
    franja = [p for p in puntos_x if abs(p[0] - punto_medio[0]) < d]
    franja.sort(key=lambda p: p[1]) # Ordenar franja por coordenada Y

    # Comparar puntos en la franja (máximo 7 comprobaciones por punto según geometría)
    for i in range(len(franja)):
        for j in range(i + 1, min(i + 7, len(franja))):
            d_franja = distancia_euclidiana(franja[i], franja[j])
            if d_franja < d:
                d, mejor_par = d_franja, (franja[i], franja[j])

    return mejor_par, d

# Prueba en celda
puntos_2d = [(1122, 6175), (135, 4076), (7296, 2741), (1120, 6170), (5000, 5000)]
print("2D Divide y Vencerás:", puntos_cercanos_2d(puntos_2d))
```

```python
import math

def puntos_cercanos_3d_general(puntos):
    min_dist = float('inf')
    par = None
    n = len(puntos)

    for i in range(n):
        for j in range(i + 1, n):
            # math.dist soporta N-dimensiones automáticamente (2D, 3D, etc.)
            d = math.dist(puntos[i], puntos[j])
            if d < min_dist:
                min_dist = d
                par = (puntos[i], puntos[j])

    return par, min_dist

# Prueba en celda (Coordenadas X, Y, Z)
puntos_3d = [(10, 20, 30), (12, 22, 31), (100, 200, 300), (11, 21, 30)]
print("3D Extension:", puntos_cercanos_3d_general(puntos_3d))
```
