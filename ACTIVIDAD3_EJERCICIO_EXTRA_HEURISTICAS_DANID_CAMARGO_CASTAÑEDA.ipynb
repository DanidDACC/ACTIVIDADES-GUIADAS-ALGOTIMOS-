```python
import random
import tsplib95

# 1. Cargar el problema localmente en esta misma celda
problem = tsplib95.load("swiss42.tsp")

# 2. Asegurar que las funciones de evaluación existen en la sesión actual
def distancia(a, b, problem):
    return problem.get_weight(a, b)

def distancia_total(solucion, problem):
    distancia_tot = 0
    for i in range(len(solucion)-1):
        distancia_tot += distancia(solucion[i], solucion[i+1], problem)
    return distancia_tot + distancia(solucion[-1], solucion[0], problem)

# 3. Crear una solución inicial aleatoria a partir de los nodos disponibles
Nodos = list(problem.get_nodes())
solucion_inicial = Nodos.copy()
random.shuffle(solucion_inicial)

# Mostrar cómo empezamos
dist_inicial = distancia_total(solucion_inicial, problem)
print(f"Distancia de la solución inicial (aleatoria): {dist_inicial}\n")

# 4. Ejecutar Búsqueda Local con Entornos Variables (VNS)
print("-" * 40)
print("Ejecutando Búsqueda Local (Entornos Variables)...")
mejor_ruta_vns, mejor_dist_vns = busqueda_local_vns(problem, solucion_inicial, max_iteraciones=5000)
print(f"Mejor distancia encontrada (VNS): {mejor_dist_vns}")
print(f"Ruta VNS: {mejor_ruta_vns}\n")

# 5. Ejecutar Recocido Simulado Mejorado
print("-" * 40)
print("Ejecutando Recocido Simulado (2-Opt)...")
mejor_ruta_rs, mejor_dist_rs = recocido_simulado_mejorado(problem, solucion_inicial, temp_inicial=10000, alfa=0.99, temp_final=0.01)
print(f"Mejor distancia encontrada (Recocido Simulado): {mejor_dist_rs}")
print(f"Ruta Recocido Simulado: {mejor_ruta_rs}")
```
